```python
from src.database.hammarby_data import query_executor, test_connection

print("Connection check:", test_connection())

# 3) Find real FK relationships in your DB (non-system schemas)
fk_sql = """
SELECT
  con.conname                               AS constraint_name,
  nsp_f.nspname                              AS fk_schema,
  rel_f.relname                              AS fk_table,
  att_f.attname                              AS fk_column,
  nsp_p.nspname                              AS pk_schema,
  rel_p.relname                              AS pk_table,
  att_p.attname                              AS pk_column,
  rn.ord                                     AS col_order
FROM pg_constraint con
JOIN pg_class rel_f       ON rel_f.oid = con.conrelid
JOIN pg_namespace nsp_f    ON nsp_f.oid = rel_f.relnamespace
JOIN pg_class rel_p       ON rel_p.oid = con.confrelid
JOIN pg_namespace nsp_p    ON nsp_p.oid = rel_p.relnamespace
JOIN LATERAL unnest(con.conkey)  WITH ORDINALITY AS fk(attnum, ord) ON TRUE
JOIN LATERAL unnest(con.confkey) WITH ORDINALITY AS pk(attnum, ord) ON fk.ord = pk.ord
JOIN pg_attribute att_f    ON att_f.attrelid = rel_f.oid AND att_f.attnum = fk.attnum
JOIN pg_attribute att_p    ON att_p.attrelid = rel_p.oid AND att_p.attnum = pk.attnum
JOIN LATERAL (SELECT fk.ord) rn ON TRUE
WHERE con.contype = 'f'
  AND nsp_f.nspname NOT IN ('pg_catalog','information_schema')
  AND nsp_p.nspname NOT IN ('pg_catalog','information_schema')
ORDER BY constraint_name, col_order;
"""
fk_df = query_executor(fk_sql)
print(f"Found FK rows: {0 if fk_df is None else len(fk_df)}")

def qident(*parts):
    """Quote identifiers safely: ("schema","table") -> "schema"."table" """
    return ".".join([f'"{p}"' for p in parts])

def pick_cols(schema, table, n=4):
    sql = """
    SELECT column_name
    FROM information_schema.columns
    WHERE table_schema=%s AND table_name=%s
    ORDER BY ordinal_position
    LIMIT %s;
    """
    cols = query_executor(sql, params=(schema, table, n))
    return [r for r in (cols["column_name"].tolist() if hasattr(cols, "columns") else [])]

if fk_df is not None and len(fk_df) > 0:
    # Group by one FK and build a multi-column join if needed
    grp = fk_df.groupby("constraint_name").head(1000)  # keep shape
    first_fk = grp.iloc[0]["constraint_name"]
    g = fk_df[fk_df["constraint_name"] == first_fk].sort_values("col_order")

    fk_schema = g.iloc[0]["fk_schema"]
    fk_table  = g.iloc[0]["fk_table"]
    pk_schema = g.iloc[0]["pk_schema"]
    pk_table  = g.iloc[0]["pk_table"]

    # Build ON clauses for all columns in this FK
    on_clauses = []
    for _, r in g.iterrows():
        on_clauses.append(f'{qident("f", r["fk_column"])} = {qident("p", r["pk_column"])}')
    on_sql = " AND ".join(on_clauses)

    # Choose a few columns to display from each side
    f_cols = pick_cols(fk_schema, fk_table, 4)
    p_cols = pick_cols(pk_schema, pk_table, 4)
    f_sel  = ", ".join([qident("f", c) + f' AS "f.{c}"' for c in f_cols]) if f_cols else "f.*"
    p_sel  = ", ".join([qident("p", c) + f' AS "p.{c}"' for c in p_cols]) if p_cols else "p.*"

    join_sql = f"""
    SELECT {f_sel}{", " if f_sel and p_sel else ""}{p_sel}
    FROM {qident(fk_schema, fk_table)} AS f
    JOIN {qident(pk_schema, pk_table)} AS p
      ON {on_sql}
    LIMIT 20;
    """
    print("Running JOIN on:", f'{fk_schema}.{fk_table}  →  {pk_schema}.{pk_table}')
    res = query_executor(join_sql)
    display(res)
else:
    # No FKs → sample a user table
    tables_sql = """
    SELECT table_schema, table_name
    FROM information_schema.tables
    WHERE table_schema NOT IN ('pg_catalog','information_schema')
    ORDER BY table_schema, table_name
    LIMIT 1;
    """
    tdf = query_executor(tables_sql)
    if tdf is not None and len(tdf) > 0:
        s, t = tdf.iloc[0]["table_schema"], tdf.iloc[0]["table_name"]
        sample_sql = f'SELECT * FROM {qident(s,t)} LIMIT 20;'
        print("No FKs found. Sampling:", f"{s}.{t}")
        display(query_executor(sample_sql))
    else:
        print("No user tables found in this database.")
```

```python
session_store = {}

def get_session_state(thread_id):
    return session_store.get(thread_id, {"history": []})

def update_session_state(thread_id, session_state):
    session_store[thread_id] = session_state
```

```python

class BuildingAgent:
    def __init__(self):
        self.graph = build_building_flow_graph()

    def handle_building_query(self, last_message, messages, metadata, thread_id, _session_state=None):
        session_state = _session_state or get_session_state(thread_id)

        initial_state = {
            "thread_id": thread_id,
            "messages": messages,
            "metadata": metadata,
            "last_message": last_message,
            "session_state": session_state,
        }

        try:
            final_state = self.graph.invoke(initial_state)
        except Exception as e:
            return {
                "content": f"An error occurred while handling your building-related request: {e}",
                "agent_answered": "building",
                "classification": "building_specific"
            }

        # Save session_state back to Redis
        new_state = final_state.get("session_state", session_state)
        update_session_state(thread_id, new_state)

        response = (
            final_state.get("final_response")
            or final_state.get("response")
            or final_state.get("prompt")
            or "No output was generated."
        )

        return {
            "content": response,
            "agent_answered": "building",
            "classification": "building_specific" , 
            "response" : response
        }

```
